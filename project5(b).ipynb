```python
# =====================================================================
# STEP 1: INSTALL STREAMLIT AND CONFIGURE CLOUDFLARED TUNNEL
# =====================================================================
print("--- Step 1: Installing Streamlit and downloading cloudflared ---")
!pip install -q streamlit

# Download the latest cloudflared binary execution file
!wget -q https://github.com
!chmod +x cloudflared-linux-amd64

# Start the cloudflared tunnel processes in background
import subprocess
subprocess.Popen(["./cloudflared-linux-amd64", "tunnel", "--url", "http://localhost:8501"])
!nohup /content/cloudflared-linux-amd64 tunnel --url http://localhost:8501 &


# =====================================================================
# STEP 2: INSTALL STREAMLIT OPTION MENU FOR NAVIGATION BAR
# =====================================================================
print("\n--- Step 2: Installing Streamlit navigation menu bar ---")
!pip install -q streamlit-option-menu


# =====================================================================
# STEP 3: CREATE THE APP.PY STREAMLIT APPLICATION SOURCE CODE
# =====================================================================
print("\n--- Step 3: Generating app.py source code file ---")

with open("app.py", "w") as f:
    f.write("""# Core libraries for building web UI, handling images, and deep learning inference
import numpy as np
import streamlit as st
import tensorflow as tf
from PIL import Image
from streamlit_option_menu import option_menu

# Configure layout settings for the Web Application UI screen
st.set_page_config(page_title="Aerial Object Detector", layout="centered")

# Sidebar navigation menu layout
with st.sidebar:
    selected_page = option_menu(
        menu_title="Main Menu",
        options=["Home", "AI Detector", "Project Details"],
        icons=["house", "shield-shaded", "info-circle"],
        menu_icon="cast",
        default_index=1,
    )

if selected_page == "Home":
    st.subheader("Welcome to the Aerial Surveillance Portal")
    st.write("This Deep Learning application distinguishes between organic birds and artificial drones instantly.")

elif selected_page == "Project Details":
    st.subheader("Data Science Project Specifications")
    st.write("- **Core Architecture**: EfficientNetB0 Transfer Learning")
    st.write("- **Calculated Validation Accuracy**: **98.66%**")
    st.write("- **Optimization Methods**: Class Balancing Weights, Dropout Regularization")

elif selected_page == "AI Detector":
    st.title("🚁 Aerial Object Classification App")
    st.write("Upload an aerial surveillance photo to identify if it is a **Bird** or a **Drone**.")

    # Load your pre-saved optimal model weights file
    @st.cache_resource
    def load_my_model():
        return tf.keras.models.load_model("best_model.h5")

    try:
        model = load_my_model()
        st.success("AI Model loaded perfectly and successfully!")
    except Exception as e:
        st.error("Model file 'best_model.h5' not found. Please upload it inside your Colab storage panel.")

    # Create file uploader box interface element
    uploaded_file = st.file_uploader("Choose an aerial image file...", type=["jpg", "jpeg", "png"])

    if uploaded_file is not None:
        # Display user uploaded image layout
        image = Image.open(uploaded_file)
        st.image(image, caption="Uploaded Aerial Image Screen Output", use_container_width=True)

        st.write("\\nProcessing image classifications... Please wait.")

        # Resize image to match model input shape dimensions (224x224)
        img_resized = image.resize((224, 224))
        img_array = np.array(img_resized)

        # Preprocess and clean image color pixel arrays accurately
        if len(img_array.shape) == 2:
            img_array = np.stack((img_array,) * 3, axis=-1)
        elif len(img_array.shape) == 3 and img_array.shape[2] == 4:
            img_array = img_array[:, :, :3]
        elif len(img_array.shape) == 3 and img_array.shape[2] == 1:
            img_array = np.concatenate([img_array] * 3, axis=-1)

        # Expand tensor dimensions to setup correct input batch matrix -> (1, 224, 224, 3)
        img_tensor = np.expand_dims(img_array, axis=0)

        # Execute neural inference models and compute metrics
        prediction = model.predict(img_tensor)

        # FIXED: Convert NumPy array scalar directly to float type to prevent format errors
        raw_probability = float(prediction[0][0])

        # In our trained dataset mapping hierarchy: Bird = 0, Drone = 1
        if raw_probability > 0.5:
            label = "DRONE 🚁"
            confidence_score = raw_probability * 100
        else:
            label = "BIRD 🦅"
            confidence_score = (1.0 - raw_probability) * 100

        # Display the final classification prediction cleanly with a metrics UI badge element
        st.metric(label=f"Result Prediction Class: {label}", value=f"{confidence_score:.2f}% Confidence")
""")

print("SUCCESS: app.py generated completely and cleanly without format string errors!")


# =====================================================================
# STEP 4: RUN STREAMLIT APP IN THE BACKGROUND WITH FIXED FILENAME
# =====================================================================
print("\n--- Step 4: Running Streamlit application in the background ---")
!streamlit run /content/app.py &>/content/logs.txt &


# =====================================================================
# STEP 5: EXTRACT AND PRINT THE LIVE TUNNEL URL FOR THE WEB APP
# =====================================================================
print("\n--- Step 5: Extracting your Live Tunnel Presentation URL Link ---")
import time
time.sleep(5) # Pause briefly to ensure the tunnel prints out completely

# Automatically prints out your private trial link to launch the presentation directly
!grep -o 'https://.*\.trycloudflare.com' nohup.out | head -n 1 | xargs -I {} echo "Your tunnel url: {}"
```
